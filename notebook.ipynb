# output
based on this output i find out there is no duplicate value in this data so now i think to clean the errors in this dataset

```python
bank['branch'] = bank['branch'].str.strip().str.title()
bank['account_type'] = bank['account_type'].str.strip().str.title()
bank['txn_city'] = bank['txn_city'].str.strip().str.title()
```

# NEXT STEP
I think to solve the time errors in this dataset


```python
bank['transaction_date'] = pd.to_datetime(bank['transaction_date'], errors='coerce')
bank['transaction_time'] = pd.to_datetime(bank['transaction_time'], format='%H:%M:%S', errors='coerce').dt.time
```

```python
print("Before:", bank['transaction_date'].unique()[:50])
print("After: ", bank['transaction_date'].unique()[:50])
```

```python
print("Date dtype:", bank["transaction_date"].dtype)
print("Missing/invalid dates:", bank["transaction_date"].isna().sum())
print(bank[["transaction_date", "transaction_time"]].head())
```

# output
now the missing and invalied date and time or cleared and make it a proper date and time

```python
for col in ['branch', 'account_type', 'txn_city']:
    bank[col] = bank[col].str.strip().str.title()

```

```python

print("Branches after: ", bank['branch'].nunique())
```

```python
bank.isnull().sum()[bank.isnull().sum() > 0]
```

```python
text_cols = ['gender', 'occupation', 'txn_city']

print("Before:\n", bank[text_cols].isnull().sum())

bank[text_cols] = bank[text_cols].fillna('Unknown')

print("\nAfter:\n", bank[text_cols].isnull().sum())
```

# explination
 in this dataset there are many missing values in the gender ,occupation ,city ...... so then i use to fill the missing value by the patter of data in this dataset before fill the missing value i check the value of missing and then i check the missing value after cleaning the data .

# NEXT STEP 
NOW I am going to find the missing value in year month in this dataset

```python
bank['transaction_date'] = bank['transaction_date'].ffill()
bank['year_month'] = bank['transaction_date'].dt.to_period('M').astype(str)
```

# NEXT STEP
Now i am going to fill the missing value in the dataset by the partten of the dataset

```python
bank['age'] = bank['age'].fillna(bank['age'].median())
bank['annual_income'] = bank['annual_income'].fillna(bank['annual_income'].median())
bank['credit_limit'] = bank['credit_limit'].fillna(0)
```

# final step 
now i am going to check the data is cleaned or not by checking the value in the dataset

```python
bank = bank.dropna(subset=['amount', 'customer_id'])
```

```python
print("Final shape:", bank.shape)
bank.to_csv("banking_cleaned_final.csv", index=False)
```
